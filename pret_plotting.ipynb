```python
map = folium.Map(location=[df['latitude'].mean(), df['longitude'].mean()], zoom_start=6)
kw = {"prefix": "fa", "color": "darkred", "icon": "star"}
angle = 90
pret_icon = folium.Icon(angle=angle, **kw)

for index, location_info in df.iterrows():
    folium.Marker([location_info["latitude"], location_info["longitude"]], popup='Pret:'+location_info["address"], icon=pret_icon).add_to(map)

map
```
