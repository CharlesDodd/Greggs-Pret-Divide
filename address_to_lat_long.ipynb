```python
import numpy as np
df['latitude'] = np.nan
df['longitude'] = np.nan
print(df.head())
```

```python
# from geopy.geocoders import Nominatim
from geopy.geocoders import ArcGIS
from geopy.extra.rate_limiter import RateLimiter
import time

geolocator = ArcGIS(user_agent="my_pret_app", timeout=10)
geocode = RateLimiter(geolocator.geocode, min_delay_seconds=1.1)
df['latlong'] = df['address'].apply(lambda x: (geocode(x).latitude, geocode(x).longitude) if geocode(x) else (None, None))

# df.to_csv('/content/pret_geocoded.csv', index=False)

print(df.head())
```

```python
# df = pd.read_csv('/content/pret_geocoded.csv')

# Extract latitude and longitude from the tuples in the 'latlong' column
df['latitude'] = df['latlong'].apply(lambda x: x[0] if isinstance(x, tuple) else None)
df['longitude'] = df['latlong'].apply(lambda x: x[1] if isinstance(x, tuple) else None)

# The columns are already named 'latitude' and 'longitude' so renaming is not needed
# df.rename(columns={'lat': 'latitude', 'long': 'longitude'}, inplace=True)

df = df.drop('latlong', axis=1)


print(df.head())

# df.to_csv('/content/pret_geocoded.csv', index=False)
```
